##### Importando bibliotecas nativas

```python
import pandas as pd
from sqlalchemy import create_engine, Integer, Date
```

##### Lendo o csv

```python
df_original = pd.read_csv('..\\ETL\\jogosNintendo.csv',sep=",")
```

```python
df_original
```

##### Conexão com o MySQL

```python
usuario = "root"
senha = "root"
host = "127.0.0.1"
porta = "3306"
banco = "projeto_nintendo"

engine = create_engine(f"mysql+pymysql://{usuario}:{senha}@{host}:{porta}/{banco}")
```

##### Criar tabela no Workbench

```python
df_original.to_sql("jogos_nintendo", 
                   con=engine, 
                   if_exists='replace', 
                   index=False, 
                   dtype={ #personalizar tipos de coluna
                       "Id Pacote": Integer(),
                       "Data Envio": Date(),
                       "Quantidade Estoque": Integer()}
                    )
```

```python
resultado = pd.read_sql("SELECT `Id Pacote`, Jogo, Console, `Data Envio`, `Status Pedido`, Gerente FROM jogos_nintendo LIMIT 15", 
                        con=engine, parse_dates=['Data Envio']) #no select, os nomes compostos devem estar entre crases (SQL e Python)

resultado['Id Pacote'] = resultado['Id Pacote'].astype('Int64').astype('string')

resultado
```
